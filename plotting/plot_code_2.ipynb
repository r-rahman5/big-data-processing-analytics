```python
#bar chart for monthly q2a task 2
import calendar
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df_peak = pd.read_csv("/home/jovyan/cw/task2a_monthly_output.csv")

df['month_name'] = df['month'].apply(lambda x: calendar.month_abbr[int(x)]) #map month to names

stations = df['Start station'].unique()
months = df['month'].unique()
month_names = [calendar.month_abbr[int(y)] for y in months]
bar_width = 0.1
x = np.arrange(len(months))

plt.figure(figsize=(12,6))

#plot for each station
for i, station in enumerate(stations):
    subset = df[df['Start station']==station]
    trip_counts = [subset[subset['month']==y]['Trip Starts'].values[0] if y in subset['month'].values else 0 for y in months]
    plt.bar(x + i*bar_width, trip_counts, width=bar_width, labe=station)


plt.title("Top 10 Start Stations per Month")
plt.xlabel("Month")
plt.ylabel("Trip Starts")
plt.xticks(x + bar_width*(len(stations)-1)/2, month_names)
plt.legend(title="Station", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
#bar chart q3 task 2
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/jovyan/cw/task2_q3_output.csv")

for col in ['Trips', 'Median (s)', '90th Percentile (s)']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

#time of day order
order_time = ["Morning", "Afternoon", "Evening", "Night"]
df['time_of_day'] = pd.Categorical(df['time_of_day'], categories=order_time, ordered=True)
df=df.sort_values('time_of_day')

#plot
fig, ax_1 = plt.subplots(figsize=(10,6))
ax_2 = ax_1.twinx() #two y axis

ax_1.plot(df['time_of_day'], df['Trips'], marker='o', color='skyblue', label='Total Trips')
ax_1.set_xlabel("Time of Day")
ax_1.set_ylabel("Total Trips", color='black')
ax_1.tick_params(axis='y', labelcolor='black')
ax_1.grid(True)

#duration stats
ax_2.plot(df['time_of_day'], df['Median (s)'], marker='s', color='green', label='Median Duration (s)')
ax_2.plot(df['time_of_day'], df['90th Percentile (s)'], marker='^', color='red', label='90th Percentile Duration (s)')
ax_2.set_ylabel("Duration (s)", color='black')
ax_2.tick_params(axis='y', labelcolor='black')

lines, labels = ax_1.get_legend_labels()
lines2, labels2 = ax_2.get_legend_labels()
ax_1.legend(lines + lines2, labels + labels2, loc='upper right')

plt.title('Utilisation by Time of Day')
plt.tight_layout()
plt.show()
```

```python
#scatter q4 task 2
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/home/jovyan/cw/task2_q4_output.csv")

plt.figure(figsize=(12,5))

#plot trip count and duration
plt.plot(df['date'], df['Trip_Count'], label='Trip Count', marker='o')
plt.plot(df['date'], df['Avg_Duration_(s)'], label='Average Duration (s)', marker='o')

#highlight anomalies
anomalies = df[df['is_anomaly']==True]
plt.scatter(anomalies['date'], anomalies['Trip_Count'], color='green', s=100, label='Trip Count Anomaly', zorder=5)
plt.scatter(anomalies['date'], anomalies['Avg_Duration_(s)'], color='skyblue', s=100, label='Duration Anomaly', zorder=5)

plt.bar(df["Route"], df["Trips"], color='skyblue')
plt.title("Daily Trip Count and Avg Duration with Anomalies")
plt.xlabel("Date")
plt.ylabel("Count / Duration")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
