```python
#bar chart q2 task 1
import pandas as pd
import matplotlib.pyplot as plt

# Load Spark output CSV
df = pd.read_csv("/home/jovyan/cw/task1_q2_output.csv")

plt.figure(figsize=(8,4))
plt.bar(df['Day of Week'], df['event_count'], color='skyblue')
plt.title("GitHub events per Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Number of Events")
plt.grid(True)
plt.show()
```

```python
#scatter plot q3b task 1
import pandas as pd
import matplotlib.pyplot as plt

# Load Spark output CSV
df = pd.read_csv("/home/jovyan/cw/task1_q3b_output.csv")

plt.figure(figsize=(8,4))
plt.scatter(df["event_count"], df["distinct_actors"], alpha=0.5)
plt.title("Actor Diversity vs Repository Popularity")
plt.xlabel("Event Count")
plt.ylabel("Distinct Actors")
plt.grid(True)
plt.show()
```

```python
#line chart q5 task 1
import pandas as pd
import matplotlib.pyplot as plt

# Load Spark output CSV
df = pd.read_csv("/home/jovyan/cw/task1_q5_output.csv")

plt.figure(figsize=(8,4))
plt.plot(df["hour"], df["total_commits"], marker='o')
plt.title("GitHub Commit activity by Hour")
plt.xlabel("Hour (0-23)")
plt.ylabel("Total Commits")
plt.grid(True)
plt.show()
```
